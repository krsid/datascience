```python
def extract_episode(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True:
            data += i 
        if i == "(":
            check = True
           
```

```python
df['Title'].apply(extract_episode)
```

```python
df['Episodes'] = df['Title'].apply(extract_episode)
```

```python
def extract_timestamp(txt):
    data = ""
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1,i+20):
                data += txt[j]
            return data
```

```python
df["Timestamp"] = df["Title"].apply(extract_timestamp)
```

```python
df
```

```python
df
```

```python
#extract the number of months from column timestamp
def total_months(period):
    start, end = period.split(' - ')

    start = pd.to_datetime(start, format='%b %Y')
    end = pd.to_datetime(end, format='%b %Y')

    return (end.year - start.year) * 12 + (end.month - start.month) +1

df['Months'] = df['Timestamp'].apply(total_months)
```
