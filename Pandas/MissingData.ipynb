```python
import numpy as np
import pandas as pd
```

```python
# Finding Missing Data
```

```python
data = {
    'A' : [1,2,np.nan,4,5],
    'B' : [1,2,3,4,5],
    'C' : [1,2,3,np.nan,np.nan],
    'D' : [1, np.nan,np.nan,np.nan,5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

```python
# Removing Missing Data
```

df

```python
df
```

```python
df.dropna()
```

```python
df.dropna(thresh=4)
```

```python
df
```

Filling the Missing Values

```python
df.fillna(0)
```

```python
v = {'A' : 20, 'B': 30,'C':10,'D': 60}
df.fillna(value=v)
```
