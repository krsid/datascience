```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

# Matbplotlib Basic

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("Basic")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

```python
#Subplots in Matplotlib
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x**2,y)
plt.subplot(2,2,4)
plt.plot(y**2,x)
```

```python
#Object Oriented way
fig = plt.figure(figsize=(10,8),dpi=100)
ax1 = fig.add_axes([0.1,0.1,0.8,0.8])
ax1.plot(x,y)
ax2 = fig.add_axes([0.2,0.3,0.2,0.2])
ax2.plot(y,x)
```

```python
#Types of plots
plt.scatter(x,y)
```

```python
plt.hist(x)
```

```python
plt.boxplot(x)
```

```python
#Saving plots
alt = plt.figure()
ax = alt.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)

plt.savefig('basicplot.png')
```
